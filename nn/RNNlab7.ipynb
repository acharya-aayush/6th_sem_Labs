```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# Create a sine wave
t = np.linspace(0, 100, 1000)
data = np.sin(t)

# Prepare sliding window dataset (W = 10)
def create_sequences(data, window):
    X, y = [], []
    for i in range(len(data) - window):
        X.append(data[i:i+window])
        y.append(data[i+window])
    return np.array(X), np.array(y)

window_size = 10
X, y = create_sequences(data, window_size)

# Reshape for RNN: [Samples, Time_Steps, Features]
X = X.reshape((X.shape[0], X.shape[1], 1))

# Split into train and test
split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# Cell 2: RNN Model Construction
model = Sequential([
    SimpleRNN(32, activation='tanh', input_shape=(window_size, 1)),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')

# Cell 3: Training
print("Training RNN on sequence data...")
model.fit(X_train, y_train, epochs=10, batch_size=32, verbose=1)

# Cell 4: Prediction and B&W Visualization
y_pred = model.predict(X_test)

plt.figure(figsize=(10, 5))
# Using different line styles and widths for B&W printing
plt.plot(y_test[:100], color='black', linewidth=2, label='Actual Sine Wave')
plt.plot(y_pred[:100], color='black', linestyle='--', linewidth=1.5, label='RNN Prediction')

plt.title("Time-Series Prediction: Actual vs RNN Forecast")
plt.xlabel("Time Steps")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True, linestyle=':')
plt.show()
```
