```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.metrics import silhouette_score
# Load data (using first 2 features for visualization clarity)
data = load_iris()
X = data.data[:, [2, 3]]  # Using Petal Length and Petal Width
k = 3
def kmeans_scratch(X, k, max_iters=100):
    # 1. Randomly initialize centroids
    indices = np.random.choice(len(X), k, replace=False)
    centroids = X[indices]
    for _ in range(max_iters):
        # 2. Assign clusters based on Euclidean distance
        distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
        labels = np.argmin(distances, axis=1)
        
        # 3. Recompute centroids
        new_centroids = np.array([X[labels == i].mean(axis=0) for i in range(k)])
        
        # 4. Check for convergence
        if np.all(centroids == new_centroids):
            break
        centroids = new_centroids
        
    return centroids, labels
centroids, labels = kmeans_scratch(X, k)
inertia = np.sum([np.sum(np.square(X[labels == i] - centroids[i])) for i in range(k)])
score = silhouette_score(X, labels)

print(f"Final Inertia: {inertia:.4f}")
print(f"Silhouette Score: {score:.4f}")
plt.figure(figsize=(8, 6))
colors = ['red', 'green', 'blue']
for i in range(k):
    cluster_data = X[labels == i]
    plt.scatter(cluster_data[:, 0], cluster_data[:, 1], c=colors[i], label=f'Cluster {i+1}', alpha=0.6)

plt.scatter(centroids[:, 0], centroids[:, 1], c='black', marker='X', s=200, label='Centroids')
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("K-Means Clustering from Scratch (Iris Dataset)")
plt.legend()
plt.grid(True)
plt.show()
```
